```python
import matplotlib
import numpy as np
import pandas as pd
import seaborn as sns
import sklearn

versoes = {
    "NumPy": np.__version__,
    "pandas": pd.__version__,
    "Matplotlib": matplotlib.__version__,
    "Seaborn": sns.__version__,
    "scikit-learn": sklearn.__version__,
}
pd.Series(versoes, name="versão").to_frame()
```

## 1. Carregamento e exploração

O conjunto Iris acompanha o scikit-learn, portanto esta aula funciona sem internet.

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame.rename(columns={"target": "especie"})
df["nome_especie"] = df["especie"].map(dict(enumerate(iris.target_names)))
df.head()
```

```python
print(f"Linhas: {df.shape[0]} | Colunas: {df.shape[1]}")
print(f"Valores ausentes: {int(df.isna().sum().sum())}")
df.groupby("nome_especie").mean(numeric_only=True).round(2)
```

## 2. Visualização

```python
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="petal length (cm)",
    hue="nome_especie",
    s=80,
    ax=ax,
)
ax.set(title="Flores Iris", xlabel="Comprimento da sépala (cm)", ylabel="Comprimento da pétala (cm)")
plt.show()
```

## 3. Modelo de classificação

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split

X, y = iris.data, iris.target
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

modelo = RandomForestClassifier(n_estimators=100, random_state=42)
modelo.fit(X_treino, y_treino)
previsoes = modelo.predict(X_teste)

print(f"Acurácia: {accuracy_score(y_teste, previsoes):.1%}")
pd.DataFrame(
    confusion_matrix(y_teste, previsoes),
    index=iris.target_names,
    columns=iris.target_names,
)
```
